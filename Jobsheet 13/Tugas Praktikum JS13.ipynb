#Tugas Praktikum

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model JST (Melengkapi bagian rumpang)
model = Sequential([
    Flatten(input_shape=(28, 28)),  # Ubah gambar 28x28 menjadi vektor
    Dense(128, activation='relu'),  # Hidden layer 1
    Dense(64, activation='relu'),  # Hidden layer 2
    Dense(10, activation='softmax'),  # Output layer (10 kelas angka 0-9)
])

# 3. Kompilasi model
model.compile(
    optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']
)

# 4. Latih model
model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f'Akurasi pada data uji: {acc:.4f}')
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Bangun model dengan neuron 256 dan 128
model = Sequential([
    Flatten(input_shape=(28, 28)),  # Mengubah gambar 28x28 menjadi vektor
    Dense(256, activation='relu'),  # Hidden layer 1 (256 neuron)
    Dense(128, activation='relu'),  # Hidden layer 2 (128 neuron)
    Dense(10, activation='softmax'),  # Output layer (10 kelas angka)
])

# 3. Kompilasi model
model.compile(
    optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']
)

# 4. Latih model
model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f'Akurasi dengan 256 & 128 neuron: {acc:.4f}')
```

2. Tambahkan satu hidden layer lagi.

```python
import time
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Catat waktu mulai pelatihan
start_time = time.time()

# 2. Bangun model dengan 3 Hidden Layer (256, 128, 64)
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(256, activation='relu'),  # Hidden layer 1
    Dense(128, activation='relu'),  # Hidden layer 2
    Dense(64, activation='relu'),  # Hidden layer 3 (Tambahan)
    Dense(10, activation='softmax'),  # Output layer
])

# 3. Kompilasi model
model.compile(
    optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']
)

# 4. Latih model
model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

# Hitung durasi waktu pelatihan
train_time = time.time() - start_time

# 5. Evaluasi model
loss, acc = model.evaluate(X_test, y_test)
print(f'\nWaktu Pelatihan (3 Hidden Layers): {train_time:.2f} detik')
print(f'Akurasi dengan 3 Hidden Layer: {acc:.4f}')
```

**Analisis Perbandingan:**

* **Waktu Pelatihan:** Penambahan satu *hidden layer* ekstra sedikit meningkatkan durasi pelatihan (menjadi 57.05 detik) karena adanya tambahan parameter dan komputasi *backpropagation*.

4. Eksperimen dengan fungsi aktivasi Sigmoid vs ReLU.

```python
import time
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import to_categorical

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)


# Fungsi untuk melatih dan mengevaluasi model
fn_activation = ['relu', 'sigmoid']
results = {}

for act in fn_activation:
  print(f'\n--- Melatih Model dengan Aktivasi: {act.upper()} ---')

  start_time = time.time()

  model = Sequential([
      Flatten(input_shape=(28, 28)),
      Dense(256, activation=act),
      Dense(128, activation=act),
      Dense(10, activation='softmax'),
  ])

  model.compile(
      optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']
  )

  model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)

  train_time = time.time() - start_time
  loss, acc = model.evaluate(X_test, y_test, verbose=0)

  results[act] = {'time': train_time, 'accuracy': acc, 'loss': loss}

# Ringkasan Hasil
print('\n========================================')
print('        HASIL EKSPERIMEN RELU vs SIGMOID      ')
print('========================================')
for act, res in results.items():
  print(
      f"Aktivasi {act.upper()} -> Akurasi: {res['accuracy']*100:.2f}% |"
      f" Waktu: {res['time']:.2f} detik"
  )
print('========================================')
```
