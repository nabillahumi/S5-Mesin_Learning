# 1. Buatlah model SVM

```python
# Import library yang dibutuhkan
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# Memuat dataset cafes.csv dan reviews.csv
cafes_df = pd.read_csv('cafes.csv')
reviews_df = pd.read_csv('reviews.csv')

# Agregasi data review per kafe
reviews_agg = reviews_df.groupby('cafe_id').agg(
    avg_review_rating=('rating', 'mean'),
    total_reviews=('review_id', 'count')
).reset_index()

# Menggabungkan dataset
merged_df = pd.merge(cafes_df, reviews_agg, on='cafe_id', how='inner').dropna(subset=['rating', 'avg_review_rating', 'review_count'])

# Menyiapkan Fitur (X) dan Label Target (y)
X = merged_df[['rating', 'review_count', 'avg_review_rating', 'total_reviews', 'latitude', 'longitude']].values
y = (merged_df['rating'] >= 4.7).astype(int).values

# Normalisasi / Standardisasi fitur
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Melakukan split data dengan rasio 70:30 dan 80:20
X_train_70, X_test_70, y_train_70, y_test_70 = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
X_train_80, X_test_80, y_train_80, y_test_80 = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

a. Split data dengan rasio 70:30 dan 80:20 untuk setiap model yang akan dibangun.

 i. Membuatan model kernel linier

```python
print("--- 1.a.1. KERNEL LINIER ---")

# Training & Evaluasi rasio 70:30
model_linear_70 = SVC(kernel='linear', random_state=42)
model_linear_70.fit(X_train_70, y_train_70)
acc_linear_70 = accuracy_score(y_test_70, model_linear_70.predict(X_test_70))

# Training & Evaluasi rasio 80:20
model_linear_80 = SVC(kernel='linear', random_state=42)
model_linear_80.fit(X_train_80, y_train_80)
acc_linear_80 = accuracy_score(y_test_80, model_linear_80.predict(X_test_80))

print(f"Akurasi Kernel Linier (Split 70:30): {acc_linear_70:.4f}")
print(f"Akurasi Kernel Linier (Split 80:20): {acc_linear_80:.4f}")
```

ii. Kernel Polynomial

```python
print("--- 1.a.2. KERNEL POLYNOMIAL ---")

# Training & Evaluasi rasio 70:30
model_poly_70 = SVC(kernel='poly', degree=3, random_state=42)
model_poly_70.fit(X_train_70, y_train_70)
acc_poly_70 = accuracy_score(y_test_70, model_poly_70.predict(X_test_70))

# Training & Evaluasi rasio 80:20
model_poly_80 = SVC(kernel='poly', degree=3, random_state=42)
model_poly_80.fit(X_train_80, y_train_80)
acc_poly_80 = accuracy_score(y_test_80, model_poly_80.predict(X_test_80))

print(f"Akurasi Kernel Polynomial (Split 70:30): {acc_poly_70:.4f}")
print(f"Akurasi Kernel Polynomial (Split 80:20): {acc_poly_80:.4f}")
```

iii. kernel RBF.

```python
print("--- 1.a.3. KERNEL RBF ---")

# Training & Evaluasi rasio 70:30
model_rbf_70 = SVC(kernel='rbf', random_state=42)
model_rbf_70.fit(X_train_70, y_train_70)
acc_rbf_70 = accuracy_score(y_test_70, model_rbf_70.predict(X_test_70))

# Training & Evaluasi rasio 80:20
model_rbf_80 = SVC(kernel='rbf', random_state=42)
model_rbf_80.fit(X_train_80, y_train_80)
acc_rbf_80 = accuracy_score(y_test_80, model_rbf_80.predict(X_test_80))

print(f"Akurasi Kernel RBF (Split 70:30): {acc_rbf_70:.4f}")
print(f"Akurasi Kernel RBF (Split 80:20): {acc_rbf_80:.4f}")
```

b. Tabulasikan performansi setiap split dan kernel berdasarkan metrik akurasi.

```python
print("--- 1.b. TABULASI PERFORMANSI ---")

# Menyusun hasil ke dalam DataFrame
performance_data = [
    {'Rasio Split (Train:Test)': '70:30', 'Kernel': 'Linear', 'Akurasi': acc_linear_70},
    {'Rasio Split (Train:Test)': '70:30', 'Kernel': 'Polynomial', 'Akurasi': acc_poly_70},
    {'Rasio Split (Train:Test)': '70:30', 'Kernel': 'RBF', 'Akurasi': acc_rbf_70},
    {'Rasio Split (Train:Test)': '80:20', 'Kernel': 'Linear', 'Akurasi': acc_linear_80},
    {'Rasio Split (Train:Test)': '80:20', 'Kernel': 'Polynomial', 'Akurasi': acc_poly_80},
    {'Rasio Split (Train:Test)': '80:20', 'Kernel': 'RBF', 'Akurasi': acc_rbf_80},
]

df_tabulasi = pd.DataFrame(performance_data)
print("\nTabel Perbandingan Performa Akurasi:")
display(df_tabulasi)
```

2. Klasifikasi Menggunakan Fitur Statistik/Histogram Teks & SVM RBF Kernel

```python
print("--- NOMOR 2: KLASIFIKASI DENGAN FITUR HISTOGRAM TEKS & SVM RBF ---")

from sklearn.model_selection import GridSearchCV

# 1. Load & Siapkan data dari reviews.csv
reviews_clean = reviews_df.dropna(subset=['text', 'rating']).copy()
reviews_clean['label'] = (reviews_clean['rating'] >= 4.8).astype(int)

# 2. Ekstraksi fitur statistik / distribusi panjang teks (sebagai fitur histogram ulasan)
reviews_clean['word_count'] = reviews_clean['text'].apply(lambda x: len(str(x).split()))
reviews_clean['char_count'] = reviews_clean['text'].apply(lambda x: len(str(x)))

X_t2 = reviews_clean[['word_count', 'char_count']].values
y_t2 = reviews_clean['label'].values

# Standardisasi fitur
scaler_t2 = StandardScaler()
X_t2_scaled = scaler_t2.fit_transform(X_t2)

# 3. Split data dengan rasio 80:20
X_train_t2, X_test_t2, y_train_t2, y_test_t2 = train_test_split(X_t2_scaled, y_t2, test_size=0.2, random_state=42)

# 4. Hyperparameter Tuning menggunakan GridSearchCV untuk Kernel RBF
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': ['scale', 'auto', 0.01, 0.1, 1]
}

grid_t2 = GridSearchCV(SVC(kernel='rbf', random_state=42), param_grid, cv=5)
grid_t2.fit(X_train_t2, y_train_t2)

# 5. Prediksi dan Evaluasi Akurasi
y_pred_t2 = grid_t2.predict(X_test_t2)
acc_t2 = accuracy_score(y_test_t2, y_pred_t2)

print(f"Hyperparameter Terbaik : {grid_t2.best_params_}")
print(f"Akurasi Model pada Test Data (80:20): {acc_t2:.4f} ({acc_t2 * 100:.2f}%)")
```
