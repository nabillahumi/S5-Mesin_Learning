#Langkah 0 - Instal & Import Library

```python
import pandas as pd
import re

print("✅ Library berhasil dimuat!")
```

#Langkah 1 - Load Data Teks Ulasan

```python
# Memuat dataset ulasan
df_reviews = pd.read_csv('reviews.csv')

# Menampilkan 5 baris pertama teks ulasan (data tidak terstruktur)
print("=== Contoh Data Teks Tidak Terstruktur ===")
df_reviews[['text']].head()
```

#Langkah 2 - Ekstraksi Fitur Teks (Padanan Ekstraksi Fitur Histogram Citra)

```python
# Membersihkan teks dari missing value
df_reviews['clean_text'] = df_reviews['text'].fillna('').astype(str).str.lower()

# 1. Ekstraksi Fitur Panjang Karakter dan Jumlah Kata
df_reviews['text_len'] = df_reviews['clean_text'].str.len()
df_reviews['word_count'] = df_reviews['clean_text'].apply(lambda x: len(x.split()))

# 2. Ekstraksi Fitur Karakter (Padanan Histogram RGB pada Citra)
# Kita buat fungsi untuk menghitung frekuensi kemunculan huruf a-z (histogram karakter)
def get_char_histogram(text):
    # Hanya menghitung huruf abjad a-z
    letters = [ch for ch in text if ch.isalpha()]
    # Buat array histogram ukuran 26 (mewakili huruf a sampai z)
    hist = [letters.count(chr(code)) for code in range(ord('a'), ord('z') + 1)]
    return hist

# Terapkan fungsi histogram karakter
df_reviews['char_histogram'] = df_reviews['clean_text'].apply(get_char_histogram)

# Cek hasil ekstraksi fitur untuk ulasan pertama
print(f"Panjang Karakter Teks Pertama: {df_reviews['text_len'].iloc[1]}")
print(f"Jumlah Kata Teks Pertama: {df_reviews['word_count'].iloc[1]}")
print(f"Panjang Array Histogram Karakter (A-Z): {len(df_reviews['char_histogram'].iloc[1])}")
print(f"Contoh Nilai Histogram Karakter (A-Z):\n{df_reviews['char_histogram'].iloc[1]}")

# Menampilkan dataframe hasil ekstraksi fitur teks
df_reviews[['text', 'text_len', 'word_count']].head()
```
