#Langkah 0 - Load Library

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

#Langkah 1 - Load Data

```python
# Memuat dataset
df_cafes = pd.read_csv('cafes.csv')
df_reviews = pd.read_csv('reviews.csv')

# Agregasi data ulasan
df_reviews['text_len'] = df_reviews['text'].fillna('').str.len()
agg_reviews = df_reviews.groupby('cafe_id').agg(
    avg_review_rating=('rating', 'mean'),
    total_likes=('likes', 'sum'),
    extracted_review_count=('review_id', 'count')
).reset_index()

# Menggabungkan kedua dataset
df = pd.merge(df_cafes, agg_reviews, on='cafe_id', how='left')

# Tampilkan data teratas hasil gabungan
df.head(3)
```

#Langkah 2 - Pre-Processing: Encoding Kolom Kategorikal (categories)

```python
# Inisialisasi LabelEncoder untuk kolom kategorikal 'categories'
le = LabelEncoder()
df['categories_encoded'] = le.fit_transform(df['categories'].astype(str))

print("=== Hasil Setelah Encoding 'categories' ===")
print(df[['categories', 'categories_encoded']].head(), "\n")
```

#Langkah 3 - Verifikasi Hasil Encoding categories

```python
print("=== Verifikasi Kolom Categories Encoded ===")
df[['name', 'categories', 'categories_encoded']].head()
```

#Langkah 4 - Pre-Processing: Imputasi Data pada Kolom phone / website

```python
# Imputasi missing value pada kolom 'phone'
df['phone'] = df['phone'].fillna('Tidak Ada')

print("=== Hasil Setelah Imputasi Missing Value pada 'phone' ===")
print(df[['name', 'phone']].head(5), "\n")
```

#Langkah 5 - Verifikasi Hasil Imputasi Data

```python
print("=== Verifikasi Jumlah Missing Value 'phone' setelah Imputasi ===")
print(df['phone'].isnull().sum())
```

#Langkah 6 - Pre-Processing: Standarisasi Fitur Numerik (review_count & rating)

```python
scaler = StandardScaler()
df['review_count_scaled'] = scaler.fit_transform(df[['review_count']])

print("=== Hasil Setelah Standardisasi 'review_count' ===")
print(df[['review_count', 'review_count_scaled']].head(10), "\n")
```

#Langkah 7 - Verifikasi Hasil Standarisasi

```python
print("=== Verifikasi 5 Baris Teratas Kolom Terdandardisasi ===")
df[['name', 'review_count', 'review_count_scaled']].head()
```

#Langkah 8 - Hasil Setelah Pre-Processing

```python
print("=== Hasil Akhir Pre-Processing (5 Baris Pertama) ===")
df[['name', 'rating', 'categories_encoded', 'phone', 'review_count_scaled']].head()
```

#Langkah 9 - Seleksi Fitur

```python
# Memilih fitur-fitur penting yang akan digunakan
selected_features = [
    'rating',
    'review_count',
    'categories_encoded',
    'avg_review_rating',
    'total_likes',
    'extracted_review_count',
    'review_count_scaled'
]

df_selected = df[selected_features]

print("=== Hasil Seleksi Fitur ===")
df_selected.head()
```
