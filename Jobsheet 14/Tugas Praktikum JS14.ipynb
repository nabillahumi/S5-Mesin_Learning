1. Modifikasi model CNN pada praktikum 2 sehingga didapatkan akurasi testing lebih dari 80%.

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# 1. Load dan Normalisasi Dataset CIFAR-10
(train_images, train_labels), (test_images, test_labels) = (
    datasets.cifar10.load_data()
)
train_images, test_images = train_images / 255.0, test_images / 255.0

# 2. Modifikasi Arsitektur CNN
model_mod = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),  # Neuron diperbesar menjadi 128
    layers.Dropout(0.3),  # Menambahkan Dropout untuk regularisasi
    layers.Dense(10),
])

# 3. Compile Model
model_mod.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy'],
)

# 4. Training Model dengan Epoch = 20
history_mod = model_mod.fit(
    train_images,
    train_labels,
    epochs=20,
    validation_data=(test_images, test_labels),
)

# 5. Evaluasi Hasil Akhir
test_loss, test_acc = model_mod.evaluate(test_images, test_labels)
print(f'\nAkurasi Pengujian Setelah Modifikasi: {test_acc*100:.2f}%')
```

2. Buatlah model CNN untuk klasifikasi dataset MNIST.

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# 1. Load Dataset MNIST
(train_images, train_labels), (test_images, test_labels) = (
    datasets.mnist.load_data()
)

# 2. Preprocessing: Normalisasi dan Reshape dimensi citra (menambahkan channel 1 untuk grayscale)
train_images = train_images.reshape((60000, 28, 28, 1)) / 255.0
test_images = test_images.reshape((10000, 28, 28, 1)) / 255.0

# 3. Membangun Arsitektur CNN untuk MNIST
model_mnist = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax'),  # Menggunakan softmax untuk 10 kelas
])

# 4. Compile Model
model_mnist.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'],
)

print('Arsitektur Model MNIST CNN:')
model_mnist.summary()

# 5. Training Model (Epoch cukup 5 karena MNIST relatif mudah dipelajari)
history_mnist = model_mnist.fit(
    train_images,
    train_labels,
    epochs=5,
    validation_data=(test_images, test_labels),
)

# 6. Evaluasi Akurasi MNIST
test_loss_m, test_acc_m = model_mnist.evaluate(test_images, test_labels)
print(f'\nAkurasi Pengujian Dataset MNIST: {test_acc_m*100:.2f}%')
```
