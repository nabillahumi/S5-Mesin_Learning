Buatlah model klasfikasi Multinomial Naive Bayes dengan ketentuan,

1. Menggunakan data spam.csv

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load data spam.csv (pastikan sudah di-drop kolom kosong & di-encode labelnya seperti sebelumnya)
df = pd.read_csv('spam.csv', encoding='latin-1')
df = df.drop(df.iloc[:, 2:], axis=1)
df.columns = ['Labels', 'SMS']
df['Labels'] = df['Labels'].map({'spam': 1, 'ham': 0})

X = df['SMS'].values
y = df['Labels'].values

# Split data training dan testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
```

2. Fitur CountVectorizer dengan mengaktifkan stop_words

```python
from sklearn.feature_extraction.text import CountVectorizer

# Inisialisasi CountVectorizer dengan stop_words='english'
cv = CountVectorizer(stop_words='english')

# Fitting dan transform data training & testing
X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)
```

3. Evaluasi hasilnya

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Train model
mnb_cv = MultinomialNB()
mnb_cv.fit(X_train_cv, y_train)

# Prediksi & Evaluasi
y_pred_cv = mnb_cv.predict(X_test_cv)
acc_cv = accuracy_score(y_test, y_pred_cv)

print("=== EVALUASI COUNTVECTORIZER ===")
print(f"Akurasi: {acc_cv:.4f}")
print(classification_report(y_test, y_pred_cv))
```

Buatlah model klasfikasi Multinomial Naive Bayes dengan ketentuan,

1. Menggunakan data spam.csv

```python
X_train_tf, X_test_tf, y_train_tf, y_test_tf = train_test_split(X, y, test_size=0.2, random_state=50)
```

2. Fitur TF-IDF dengan mengaktifkan stop_words

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Inisialisasi TfidfVectorizer dengan stop_words='english'
tfidf = TfidfVectorizer(stop_words='english')

# Fitting dan transform data training & testing
X_train_tfidf = tfidf.fit_transform(X_train_tf)
X_test_tfidf = tfidf.transform(X_test_tf)
```

3. Evaluasi hasilnya dan bandingkan dengan hasil pada Tugas no 2

```python
# Train model
mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train_tf)

# Prediksi & Evaluasi
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test_tf, y_pred_tfidf)

print("=== EVALUASI TF-IDF ===")
print(f"Akurasi: {acc_tfidf:.4f}")
print(classification_report(y_test_tf, y_pred_tfidf))

print("\n========================================")
print(f"PERBANDINGAN AKURASI:")
print(f"1. CountVectorizer : {acc_cv:.4f}")
print(f"2. TF-IDF          : {acc_tfidf:.4f}")
print("========================================")
```

**Fitur Terbaik**: TF-IDF cenderung lebih baik karena memberikan bobot lebih tinggi pada kata-kata yang khas dan mengurangi pengaruh kata-kata yang terlalu sering muncul. Dengan begitu, model lebih mudah mengenali kata-kata yang menjadi ciri pesan spam.
