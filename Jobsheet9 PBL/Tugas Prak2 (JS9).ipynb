#Persiapan Data Awal

#Bagian 1: Multinomial Naive Bayes dengan CountVectorizer (Stop Words Diaktifkan)

1. Menggunakan Dataset Teks Ulasan

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Load data reviews
reviews = pd.read_csv('reviews.csv')

# Menyiapkan DataFrame teks dan label
df = reviews[['text', 'rating']].copy()
df.columns = ['SMS', 'Labels'] # Penyesuaian nama kolom sesuai struktur praktikum

# Encode label (Rating >= 4.0 sebagai 1 / Positif, sisanya 0 / Negatif)
df['Labels'] = df['Labels'].apply(lambda x: 1 if x >= 4.0 else 0)

X = df['SMS'].values
y = df['Labels'].values

# Split data training dan testing (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
```

2. Kafe & Fitur CountVectorizer (stop_words='english')

```python
# Inisialisasi CountVectorizer dengan mengaktifkan stop_words
cv = CountVectorizer(stop_words='english')

# Fitting dan transform pada data training, transform pada data testing
X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)
```

3. Training & Evaluasi Model CountVectorizer

```python
# Inisialisasi dan latih model MultinomialNB
mnb_cv = MultinomialNB()
mnb_cv.fit(X_train_cv, y_train)

# Prediksi dan evaluasi
y_pred_cv = mnb_cv.predict(X_test_cv)
acc_cv = accuracy_score(y_test, y_pred_cv)

print(f"Akurasi Model (CountVectorizer + Stop Words): {acc_cv:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_cv))
```

#Bagian 2: Multinomial Naive Bayes dengan TF-IDF (Stop Words Diaktifkan)

1. 1 & 2. Menggunakan Fitur TF-IDF dengan stop_words='english'

```python
# Inisialisasi TfidfVectorizer dengan mengaktifkan stop_words
tfidf = TfidfVectorizer(stop_words='english')

# Fitting dan transform pada data training, transform pada data testing
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

3. Training, Evaluasi, dan Perbandingan Model TF-IDF

```python
# Inisialisasi dan latih model MultinomialNB menggunakan TF-IDF
mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train)

# Prediksi dan evaluasi
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)

print(f"Akurasi Model (TF-IDF + Stop Words): {acc_tfidf:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_tfidf))

print("\n--- Perbandingan Akurasi ---")
print(f"CountVectorizer : {acc_cv:.4f}")
print(f"TF-IDF          : {acc_tfidf:.4f}")
```

4. Berikan kesimpulan fitur mana yang terbaik

Berdasarkan hasil perbandingan di atas, CountVectorizer memberikan performa yang lebih baik dengan akurasi 0.9283 (92.83%) dibandingkan TF-IDF yang menghasilkan akurasi 0.8924 (89.24%). Oleh karena itu, pada kasus teks ulasan kafe ini, ekstraksi fitur menggunakan CountVectorizer lebih optimal dalam membantu model Naive Bayes mengenali pola data.
