#Langkah 1: Load Data

```python
# Load data cafes dan reviews
import numpy as np
import pandas as pd

cafes = pd.read_csv('cafes.csv')
reviews = pd.read_csv('reviews.csv', encoding='latin-1')

# Tampilkan beberapa baris awal dari data reviews
print("Data Reviews:")
display(reviews.head())
```

#Langkah 2: Preprocessing

```python
# 1. Pilih kolom yang relevan dari reviews (misal: 'text' dan 'rating')
df = reviews[['text', 'rating']].copy()
df.columns = ['SMS', 'Labels'] # Menyesuaikan penamaan dengan struktur praktikum

# 2. Inspeksi Data
print("Jumlah Data Per Kelas:")
print(df['Labels'].value_counts())
print('\nInformasi Data:')
print(df.info())

# 3. Encode Label (Contoh: Rating >= 4.0 diberi label 1 / Positif, < 4.0 diberi label 0 / Negatif)
df['Labels'] = df['Labels'].apply(lambda x: 1 if x >= 4.0 else 0)

print("\nSetelah Encoding Label:")
display(df.head())

# 4. Pisahkan Fitur (Teks) dengan Label
X = df['SMS'].values
y = df['Labels'].values
```

#Langkah 3: Ekstraksi Fitur (Bag of Words)

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Split data training dan testing (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Inisiasi CountVectorizer
bow = CountVectorizer()

# Fitting dan transform X_train dengan CountVectorizer
X_train = bow.fit_transform(X_train)

# Transform X_test (hanya transform, tidak fit, agar data testing tetap asing bagi model)
X_test = bow.transform(X_test)

# Cek dimensi fitur dari proses CountVectorizer
print(f"Jumlah Fitur Kata (Vocabulary): {len(bow.get_feature_names_out())}")
print(f'Dimensi data training: {X_train.shape}')
```

#Langkah 4: Training dan Evaluasi Model (Multinomial Naive Bayes)

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Inisiasi MultinomialNB
mnb = MultinomialNB()

# Fit model pada data training
mnb.fit(X_train, y_train)

# Prediksi dengan data training
y_pred_train = mnb.predict(X_train)
acc_train = accuracy_score(y_train, y_pred_train)

# Prediksi dengan data testing
y_pred_test = mnb.predict(X_test)
acc_test = accuracy_score(y_test, y_pred_test)

# Print hasil evaluasi
print(f'Hasil akurasi data train: {acc_train:.4f}')
print(f'Hasil akurasi data test: {acc_test:.4f}')
```
