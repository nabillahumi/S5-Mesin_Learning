#Langkah 1: Load Dataset & Preprocessing untuk Regresi Keras

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf

# Memuat dataset cafes.csv dan reviews.csv
cafes_df = pd.read_csv('cafes.csv').dropna(subset=['rating', 'review_count'])

# Menggunakan 'review_count' sebagai fitur (X) dan 'rating' sebagai target regresi (y)
X = cafes_df[['review_count']].values
y = cafes_df[['rating']].values

# Normalisasi / Standardisasi data X dan y
scaler_X = StandardScaler()
scaler_y = StandardScaler()
X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y)

# Split data (80:20)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

print("Shape X_train:", X_train.shape)
print("Shape y_train:", y_train.shape)
```

#Langkah 2: Membangun & Melatih Model Regresi Keras (Konfigurasi Awal)

```python
print("--- PRAKTIKUM 3: MODEL REGRESI DASAR ---")

# Membangun arsitektur model regresi
model_awal = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(1)
])

# Kompilasi model dengan optimizer Adam default
model_awal.compile(optimizer='adam', loss='mse')

# Latih model
history_awal = model_awal.fit(X_train, y_train, epochs=100, verbose=0)

# Evaluasi / Prediksi
pred_awal = model_awal.predict(X_test)
print("Contoh Prediksi (Skala Terstandarisasi):")
print(pred_awal[:5].flatten())
```

#Tugas 4
1.  Mengubah Learning Rate (Eksperimen LR Berbeda)

```python
print("--- TUGAS 4.1: UBAH LEARNING RATE ---")

from tensorflow.keras.optimizers import Adam

# Model dengan Learning Rate Kecil (1e-4)
model_lr_kecil = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(1)
])
model_lr_kecil.compile(optimizer=Adam(learning_rate=1e-4), loss='mse')
history_lr_kecil = model_lr_kecil.fit(X_train, y_train, epochs=100, verbose=0)

# Model dengan Learning Rate Besar (1e-2)
model_lr_besar = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(1)
])
model_lr_besar.compile(optimizer=Adam(learning_rate=1e-2), loss='mse')
history_lr_besar = model_lr_besar.fit(X_train, y_train, epochs=100, verbose=0)

print("Pelatihan model dengan berbagai Learning Rate selesai!")
```

2. Membandingkan Hasil Loss Antar Learning Rate

```python
print("--- TUGAS 4.2: PERBANDINGAN HASIL LOSS ---")

loss_awal_val = model_awal.evaluate(X_test, y_test, verbose=0)
loss_kecil_val = model_lr_kecil.evaluate(X_test, y_test, verbose=0)
loss_besar_val = model_lr_besar.evaluate(X_test, y_test, verbose=0)

print("=== TABEL PERBANDINGAN LOSS BERDASARKAN LEARNING RATE ===")
print(f"{'Konfigurasi Optimizer':<25} | {'Test Loss (MSE)':<15}")
print("-" * 45)
print(f"{'Adam (Default / 1e-3)':<25} | {loss_awal_val:.6f}")
print(f"{'Adam (LR Kecil / 1e-4)':<25} | {loss_kecil_val:.6f}")
print(f"{'Adam (LR Besar / 1e-2)':<25} | {loss_besar_val:.6f}")
print("\nAnalisis: Learning rate yang terlalu kecil membuat konvergensi lambat, sedangkan learning rate yang terlalu besar dapat menyebabkan bobot melompat melewati titik optimal sehingga loss akhir kurang stabil.")
```
