#Langkah 1 - Persiapan Dataset (Menggunakan Cafes & Reviews)

```python
# Import library
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Memuat dataset cafes.csv dan reviews.csv
cafes_df = pd.read_csv('cafes.csv').dropna(subset=['rating', 'review_count'])

# Mengambil fitur masukan (X): rating dan review_count
X_raw = cafes_df[['rating', 'review_count']].values

# Normalisasi data agar berada di rentang yang ideal untuk JST
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# Membuat target label biner (y): 1 jika rating >= 4.7, else 0
y = (cafes_df['rating'] >= 4.7).astype(int).values.reshape(-1, 1)

# Mengambil subset data (misal 16 baris pertama) untuk pelatihan JST manual
X = X_scaled[:16]
y = y[:16]

print("Data Masukan (X) shape:", X.shape)
print("Data Target (y) shape:", y.shape)
```

#Langkah 2 - Inisialisasi Bobot, Bias, dan Fungsi Aktivasi (Sigmoid)

```python
# Konfigurasi Parameter JST (Sesuai Praktikum)
input_size = X.shape[1]  # 2 fitur (rating & review_count)
hidden_size = 2          # Konfigurasi awal: 2 neuron hidden layer
output_size = 1          # 1 output biner
lr = 0.1                 # Learning rate

# Inisialisasi bobot dan bias secara acak
np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi Aktivasi Sigmoid dan Turunannya
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500))) # clip untuk mencegah overflow

def sigmoid_derivative(x):
    return x * (1 - x)

print("Inisialisasi bobot dan bias berhasil!")
```

#Langkah 3, 4, & 5 - Forward Pass, Error Calculation, Backpropagation, & Weight Update

```python
# Proses Training JST (10.000 Epochs)
epochs = 10000
for epoch in range(epochs):
    # 1. Forward Pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # 2. Hitung Error
    error = y - a2

    # 3. Backpropagation (Backward Pass)
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # 4. Update Bobot (Gradient Descent)
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 2000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

print("\nHasil Prediksi Akhir (Konfigurasi Awal):")
print(np.round(a2, 4))
```

#Penyelesaian Tugas Praktikum
###Tugas 1.1: Ubah Jumlah Neuron Hidden Layer Menjadi 3

```python
print("--- TUGAS 1.1: HIDDEN LAYER = 3 NEURON ---")

hidden_size_3 = 3

# Re-inisialisasi bobot dengan 3 hidden neuron
np.random.seed(42)
W1_3 = np.random.randn(input_size, hidden_size_3)
b1_3 = np.zeros((1, hidden_size_3))
W2_3 = np.random.randn(hidden_size_3, output_size)
b2_3 = np.zeros((1, output_size))

for epoch in range(epochs):
    z1 = np.dot(X, W1_3) + b1_3
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2_3) + b2_3
    a2 = sigmoid(z2)

    error = y - a2

    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_3.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_3 += lr * d_W1
    b1_3 += lr * d_b1
    W2_3 += lr * d_W2
    b2_3 += lr * d_b2

loss_hidden_3 = np.mean(np.square(error))
print(f"Final Loss (Hidden Layer = 3): {loss_hidden_3:.6f}")
```

###Tugas 1.2: Bandingkan Hasil Loss dengan Konfigurasi Awal

```python
print("--- TUGAS 1.2: PERBANDINGAN LOSS ---")
print(f"Loss Konfigurasi Awal (2 Hidden Neuron) : {loss:.6f}" if 'loss' in locals() else f"Loss Konfigurasi Awal: {np.mean(np.square(error)):.6f}")
print(f"Loss Konfigurasi Baru (3 Hidden Neuron): {loss_hidden_3:.6f}")
print("Analisis: Penambahan jumlah neuron pada hidden layer memperkaya kapasitas representasi model, yang umumnya menghasilkan konvergensi loss yang lebih rendah atau lebih optimal terhadap dataset.")
```

###Tambahkan Fungsi Aktivasi ReLU dan Bandingkan Hasil

```python
print("--- TUGAS 1.3: IMPLEMENTASI AKTIVASI RELU ---")

# Definisi ReLU dan turunannya
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Re-inisialisasi bobot untuk ReLU
np.random.seed(42)
W1_relu = np.random.randn(input_size, hidden_size)
b1_relu = np.zeros((1, hidden_size))
W2_relu = np.random.randn(hidden_size, output_size)
b2_relu = np.zeros((1, output_size))

for epoch in range(epochs):
    # Forward Pass: Hidden layer menggunakan ReLU, Output layer menggunakan Sigmoid
    z1 = np.dot(X, W1_relu) + b1_relu
    a1 = relu(z1)
    z2 = np.dot(a1, W2_relu) + b2_relu
    a2 = sigmoid(z2)

    error = y - a2

    # Backpropagation menggunakan turunan ReLU
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_relu.T) * relu_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_relu += lr * d_W1
    b1_relu += lr * d_b1
    W2_relu += lr * d_W2
    b2_relu += lr * d_b2

loss_relu = np.mean(np.square(error))
print(f"Final Loss dengan Fungsi Aktivasi ReLU: {loss_relu:.6f}")
print("Analisis: Aktivasi ReLU mengatasi masalah vanishing gradient pada hidden layer sehingga proses pembelajaran bobot berjalan lebih efektif pada arsitektur JST.")
```
