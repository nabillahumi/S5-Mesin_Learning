1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
import pandas as pd

# Load dataset cafe
df_cafes = pd.read_csv('cafes.csv')

# 1. Menentukan kolom yang tidak digunakan (drop) dan kolom yang digunakan
cols_to_drop = ['cafe_id', 'name', 'address', 'phone', 'website', 'scraped_at']
df_filtered = df_cafes.drop(columns=cols_to_drop)

# Tampilkan 5 data teratas dari hasil pemilihan variabel
df_filtered.head()
```

2. Lakukan proses encoding pada kolom "diagnosis".

```python
from sklearn.preprocessing import LabelEncoder

# Membuat objek LabelEncoder
le = LabelEncoder()

# Mendeteksi kolom kategorikal yang tersisa
cat_cols = df_filtered.select_dtypes(include=['object']).columns.tolist()

# Melakukan proses encoding untuk setiap kolom kategorikal
for col in cat_cols:
    df_filtered[col] = le.fit_transform(df_filtered[col])

# Menampilkan 5 data teratas setelah encoding
df_filtered.head()
```

3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
from sklearn.preprocessing import StandardScaler

# Membuat objek StandardScaler
std = StandardScaler()

# Mendeteksi semua kolom yang bertipe numerik (int64 dan float64)
num_cols = df_filtered.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Melakukan standardisasi pada seluruh kolom numerik
df_filtered[num_cols] = std.fit_transform(df_filtered[num_cols])

# Menampilkan 5 data teratas setelah standardisasi
df_filtered.head()
```
