#Langkah 0 - Persiapan

```python
import pandas as pd
import numpy as np
```

#Langkah 1 - Memuat Data

```python
# Memuat data
df_cafes = pd.read_csv('cafes.csv')
df_reviews = pd.read_csv('reviews.csv')

# Tampilkan 5 data teratas dari dataset cafe
df_cafes.head()
```

#Langkah 2 - Inspeksi Data (Pengecekan Missing Value)

```python
# Cek missing value pada dataset cafe
print("Missing values pada df_cafes:")
print(df_cafes.isnull().sum())

print("\nMissing values pada df_reviews:")
print(df_reviews.isnull().sum())
```

#Langkah 3 - Data Imputation

```python
# Imputasi kolom 'phone' dengan keterangan "Tidak Ada"
df_cafes['phone'].fillna(value='Tidak Ada', inplace=True)

# Imputasi kolom 'website' dengan keterangan "Tidak Ada"
df_cafes['website'].fillna(value='Tidak Ada', inplace=True)
```

#Langkah 4 - Validasi Hasil

```python
# Cek kembali jumlah missing value setelah imputasi
print("Missing values setelah imputasi pada df_cafes:")
print(df_cafes.isnull().sum())

# Cek struktur informasi dataframe
df_cafes.info()
```
