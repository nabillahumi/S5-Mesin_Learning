#Langkah 0 - Persiapan

```python
import pandas as pd
import re
```

#Langkah 1 - Memuat Data Teks Ulasan

```python
# Memuat dataset review
df_reviews = pd.read_csv('reviews.csv')

# Menampilkan beberapa baris pertama pada kolom 'text' (data tidak terstruktur)
df_reviews['text'].head()
```

#Langkah 2 - Pembersihan Teks (Text Cleaning)

```python
import re

# Fungsi untuk membersihkan teks ulasan
def clean_text(text):
    if not isinstance(text, str):
        return ""
    # 1. Mengubah semua huruf menjadi huruf kecil (Lowercasing)
    text = text.lower()
    # 2. Menghapus emoji atau simbol/karakter khusus (hanya menyisakan huruf, angka, dan spasi)
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    return text

# Menerapkan fungsi ke kolom 'text' dan menyimpannya di kolom baru 'cleaned_text'
df_reviews['cleaned_text'] = df_reviews['text'].apply(clean_text)

# Menampilkan perbandingan sebelum dan sesudah dibersihkan
df_reviews[['text', 'cleaned_text']].head(5)
```

#Langkah 3 - Konversi / Ekstraksi Fitur Teks

```python
# Ekstraksi fitur sederhana: Menghitung jumlah kata dan karakter dari teks yang sudah dibersihkan
df_reviews['word_count'] = df_reviews['cleaned_text'].apply(lambda x: len(x.split()))
df_reviews['char_count'] = df_reviews['cleaned_text'].apply(len)

# Menampilkan hasil ekstraksi fitur teks
df_reviews[['cleaned_text', 'word_count', 'char_count']].head(5)
```
