#Langkah 1 - Load Data

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Memuat dataset cafe yang sudah dipersiapkan
dpath = 'cafes.csv'
df_cafes = pd.read_csv(dpath)

# Tampilkan 5 data teratas
df_cafes.head()
```

#Langkah 2 - Slice Data

```python
# Memilih subset kolom yang dianggap sebagai fitur
df_selected = df_cafes[['rating', 'review_count', 'categories', 'source_query']]

# Tampilkan 5 data teratas dari hasil slicing
df_selected.head()
```

#Langkah 3 - Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder() # Membuat objek dari LabelEncoder

# Proses encoding pada kolom kategorikal dataset kafe
df_selected['categories'] = le.fit_transform(df_selected['categories'])
df_selected['source_query'] = le.fit_transform(df_selected['source_query'])
```

#Langkah 4 - Verifikasi Hasil

```python
# Cek 5 data teratas setelah encoding
df_selected.head()
```

#Langkah 5 - Standarisasi

```python
from sklearn.preprocessing import StandardScaler

# Cek hasil encoding
df_selected.head()

# Proses standarisasi pada kolom 'review_count'
std = StandardScaler()
df_selected['review_count'] = std.fit_transform(df_selected[['review_count']])
```
