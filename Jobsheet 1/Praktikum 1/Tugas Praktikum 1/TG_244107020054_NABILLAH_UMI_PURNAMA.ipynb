**3. Jelaskan kegunakan masing-masing library tersebut.**

**Jawaban :**

PyPREP (pyprep)

Kegunaan: Library Python yang mengimplementasikan PREP (Preprocessing Pipeline) terstandarisasi untuk data Electroencephalography (EEG). PyPREP bekerja di atas ekosistem MNE-Python untuk melakukan penyaringan derau (noise reduction), identifikasi dan penanganan saluran sinyal bermasalah (bad channel detection), serta re-referencing data sinyal otak secara otomatis.

SciPy (scipy)

Kegunaan: Library dasar (core library) dalam bahasa Python yang digunakan untuk komputasi ilmiah (scientific computing) dan rekayasa. SciPy dibangun di atas NumPy dan menyediakan berbagai modul tingkat tinggi untuk optimasi numerik, aljabar linier, integrasi, transformasi Fourier, pemrosesan sinyal (signal processing), pemrosesan citra, dan analisis statistik.

wandb (wandb / Weights & Biases)

Kegunaan: Platform untuk pelacakan eksperimen Machine Learning dan Deep Learning (ML experiment tracking). Library ini digunakan untuk mencatat dan memvisualisasikan hyperparameter tuning, metrik pelatihan (seperti loss dan accuracy) secara real-time, memantau konsumsi sumber daya sistem, serta mengelola versi model (model versioning) dan dataset.

pyECG (pyecg)

Kegunaan: Library Python yang dirancang khusus untuk analisis dan pemrosesan sinyal Electrocardiogram (ECG/EKG). Library ini mempermudah pembacaan dan penulisan format dataset medis standar (seperti PhysioBank WFDB) serta membantu mengekstraksi fitur-fitur sinyal jantung untuk keperluan riset medis atau klasifikasi kesehatan.
